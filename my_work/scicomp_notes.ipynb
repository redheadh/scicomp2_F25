## Central Limit Thm

- Pick $N$ random numbers from $h(x)$ 
    - tails of PDF must fall faster than $\frac{1}{x^2}$ or its not a Cauchy distribution
- mean of those $N$ samples will converge to a Gaussian as $N \rightarrow \infty$
- This is why repeated observations imporve measurements by $\frac{1}{\sqrt{N}} \rightarrow \frac{\sigma}{\sqrt{N}}$ How you imporve your sigma

## Multivariate PDF
- characteristic PDF in more than 1 dimension or parameter
$$p(x | \mu,\sigma) = \frac{1}{\sqrt{2\pi \sigma^2}}e^{\frac{1}{2}\frac{(x-\mu)^2}{?}}$$

$$p(\vec{x} | \vec{\mu}, C_{n \times n}) = \frac{1}{\sqrt{\mathrm{det}(2\pi C}}e^{\frac{1}{2}(\vec{x}-\vec{\mu})^T C^{-1} (\vec{x} - \vec{\mu})}$$

- C is an nxn mtx with $\sigma_i ^2$ on the diagonal and 0'son the off-diagonals minus $\frac{1}{2}\frac{(x_1 - \mu_1)^2}{\sigma_1^2}$

$$C_{ij} \equiv \int_{-\infty}^{\infty} (x_i - \mu_i)(x_j - \mu_j)p(\vec{x} | \vec{\mu} C) d^nx$$

## Correlation Coefficients

    - if these are random then then sum tends towards zero
$$r = \frac{\sum_i (x_i-\bar{x})(y_i - \bar{y})}{\sqrt{(x_i-\bar{x})^2(y_i - \bar{y})^2}}$$

## Sampling from Arbitrary Distributions
- you have to know what the distributions are
- Rejection Sampling
    - say you have a distribution $h(x)$. You set some upper limit $Q(x)$. You generate a random sample or a proposal distribution and check to see if the value is below your upper limit $Q(x)$ or not. If yes, accept and go to next one. If not, reject and try again. 
    - you can use this to discern what a distribution is if you don't know or scipy doesn't know. 

- Inverse Transform Sampling
    - given some distribution $h(x)$ you can numerically find the CDF $H(x)$. Because it is normalized, the CDF will always start near 0 and then asymptote to 1. Then take uniform random samples from [0,1]  on the y-axis which will give you a collection of values for x. 
    $$H^{-1}(x) \rightarrow h(x)$$
